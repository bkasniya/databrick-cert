```sql
CREATE or REFRESH MATERIALIZED VIEW gold_customer_order_summary
AS 
select
    c.customer_id,
    c.customer_name,
    c.email,
    c.date_of_birth,
    c.telephone,
    a.address_line_1,
    a.city,
    a.state,
    a.postcode,
    count(DISTINCT o.order_id) as total_orders,
    sum(o.quantity) as total_items_orders,
    sum(o.quantity * o.price) as total_amount
from LIVE.silver_customers c join
LIVE.silver_address a on c.customer_id = a.customer_id
join LIVE.silver_orders o on c.customer_id = o.customer_id
where a.__END_AT is null
group by ALL;

```
