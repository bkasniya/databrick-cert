## Behaviour of create or replace table

```sql
CREATE OR REPLACE TABLE demo.delta.company
( company_id BigInt not null GENERATED ALWAYS as Identity (start with 1 increment by 1) ,
  company_name String not null,
Foundation Date COMMENT 'This is company foundation date', 
Foundation_year int generated always as (year(Foundation)),
Country String
)
COMMENT "this table contains company static data"
TBLPROPERTIES('sensitive'= 'true');

INSERT into demo.delta.company(company_name, Foundation, Country)
VALUES ('Microsoft', '1975-04-04','USA'),
 ('Apple', '1985-04-04','USA');

 select * from demo.delta.company;

 desc history demo.delta.company
```

## Behaviour of drop and create table

```sql
DROP TABLE if exists  demo.delta.company;

CREATE TABLE if not EXISTS demo.delta.company
( company_id BigInt not null GENERATED ALWAYS as Identity (start with 1 increment by 1) ,
  company_name String not null,
Foundation Date COMMENT 'This is company foundation date', 
Foundation_year int generated always as (year(Foundation)),
Country String
)
COMMENT "this table contains company static data"
TBLPROPERTIES('sensitive'= 'true');

INSERT into demo.delta.company(company_name, Foundation, Country)
VALUES ('Microsoft', '1975-04-04','USA'),
 ('Apple', '1985-04-04','USA');

 select * from demo.delta.company;

  desc history demo.delta.company
```
