```python
from pyspark.sql.types import *

customer_scheam = StructType(fields =[StructField("customer_id", IntegerType()),
                                     StructField("customer_name", StringType()),
                                     StructField("date_of_birth", StringType()),
                                     StructField("telephone", StringType()),
                                     StructField("email", StringType()),
                                     StructField("member_since", DateType()),
                                     StructField("created_timestamp", TimestampType())] )   
```

```python
from pyspark.sql.functions import *

df = spark.readStream.format("json").schema(customer_scheam).load("/Volumes/gizmo_box/landing/operational_data/customer_stream/").withColumn("filepath", col("_metadata.file_path")).withColumn("current_timestamp", current_timestamp())

```

```python
stream_query = (df.writeStream.format("delta").outputMode("append").option("checkpointLocation", "/Volumes/gizmo_box/landing/operational_data/checkpoints").toTable("gizmo_box.bronze.customer_stream"))
```

```python
%sql
select *  from gizmo_box.bronze.customer_stream
```
