```python
from pyspark.sql.functions import *

df = spark.readStream. \
format("cloudFiles") \
.option("cloudFiles.format", "json") \
.option("cloudFiles.schemaLocation", "/Volumes/gizmo_box/landing/operational_data/customer_autoloader/schema") \
.option("cloudFiles.schemaEvolutionMode", "addNewColumns") \
.option("cloudFiles.inferColumnTypes", "true") \
.option("cloudFiles.schemaHints", "date_of_birth DATE, member_since DATE, created_timestamp TIMESTAMP") \
.load("/Volumes/gizmo_box/landing/operational_data/customer_autoloader/") \
.withColumn("filepath", col("_metadata.file_path")) \
.withColumn("current_timestamp", current_timestamp())

```

```python
stream_query = (df.writeStream.format("delta").outputMode("append").option("checkpointLocation", "/Volumes/gizmo_box/landing/operational_data/customer_autoloader/_checkpoint").toTable("gizmo_box.bronze.customer_autoloader"))
```
