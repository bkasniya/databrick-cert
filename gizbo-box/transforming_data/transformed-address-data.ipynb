```python
from pyspark.sql.functions import *

df = spark.table("gizmo_box.bronze.v_address")
df = df.groupBy(df.customer_id).pivot('address_type').agg(max(df.address_line_1).alias('address_line_1'), max(df.city).alias('city'), max(df.state).alias('state'), max(df.postcode).alias('postcode'))

display(df)
```

```python
%sql

CREATE TABLE if not exists gizmo_box.silver.address
AS
select * from
(
select customer_id, address_type, address_line_1, city, state, postcode from gizmo_box.bronze.v_address
)
PIVOT (
max(address_line_1) as address_line_1, 
max(city) as city,
max(state) as state,
max(postcode) as postcode
for address_type in ('shipping', 'billing'))
;


```
