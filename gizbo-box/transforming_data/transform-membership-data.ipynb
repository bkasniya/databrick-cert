```python
from pyspark.sql.functions import *

df= spark.table("gizmo_box.bronze.v_membership")

# df = df.withColumn("image_name", regexp_replace("path", ".*/", ""))\
#     .withColumn("customer_id", regexp_replace("image_name", "\.png$", "")) \
#         .drop("image_name").drop("path")

df = df.withColumn("customer_id", regexp_extract(col("path"), ".*/([0-9]+)\\.png$", 1)) \
    .withColumnRenamed("content", "membercard") \
        .select(col("customer_id"), col("membercard"))


display(df)
```

```python
%sql
CREATE TABLE If not exists gizmo_box.silver.membership as
SELECT
  content as membercard,
  regexp_extract(path, "./*([0-9]+)\\.png$", 1) as customer_id
from
  gizmo_box.bronze.v_membership;
```
