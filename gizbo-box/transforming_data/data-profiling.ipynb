## Manual Data Profiling

## COUNT

```python
%sql
SELECT count(*) as total_record, count(customer_id) customer_id_count, count(email) email_count, count(telephone) telephone_count
from gizmo_box.bronze.v_customer
```

```python
%sql
SELECT count(*) as total_record, count_if(customer_id is NULL) customer_null_count, count_if(email is NULL) email_null_count, count_if(telephone is null) telephone_null_count
from gizmo_box.bronze.v_customer
```

```python
%sql
select count(distinct customer_id) from gizmo_box.bronze.v_customer where customer_id is not null
```
