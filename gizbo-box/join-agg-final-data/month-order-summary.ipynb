```python
%sql

CREATE OR REPLACE TEMPORARY VIEW order_month_data 
as
select customer_id, item_id,order_id, price, quantity, date_format(transaction_timestamp, 'yyyy-MM') as order_month from gizmo_box.silver.orders;


CREATE TABLE if not exists gizmo_box.gold.order_monthly_summary
as
select customer_id, order_month, count(DISTINCT order_id) total_orders, sum(quantity) as total_items_brought, sum(price*quantity) as total_amount  from order_month_data group by customer_id, order_month



```
