## Create Table and insert data using insert Into

```sql
CREATE TABLE if not exists demo.delta.company_write
( company_id BigInt not null GENERATED ALWAYS as Identity (start with 1 increment by 1) ,
  company_name String not null,
Foundation Date COMMENT 'This is company foundation date', 
Foundation_year int generated always as (year(Foundation)),
Country String
)
COMMENT "this table contains company static data"
TBLPROPERTIES('sensitive'= 'true');

INSERT into demo.delta.company_write(company_name, Foundation, Country)
VALUES ('Microsoft', '1975-04-04','USA'),
 ('Apple', '1985-04-04','USA');

 select * from demo.delta.company_write;

INSERT overwrite demo.delta.company_write(company_name, Foundation, Country)
VALUES ('Microsoft', '1975-04-04','USA'),
 ('Apple', '1985-04-04','USA');

 select * from demo.delta.company_write;

 desc history demo.delta.company_write
```

## Overwrite data by partition

```sql
DROP TABLe if exists  demo.delta.company_write_part;

CREATE TABLE if not exists demo.delta.company_write_part
( company_name String not null,
Foundation Date COMMENT 'This is company foundation date', 
Country String
)
COMMENT "this table contains company static data"
TBLPROPERTIES('sensitive'= 'true')
PARTITIONED BY (company_name) ;

INSERT into demo.delta.company_write_part(company_name, Foundation, Country)
VALUES ('Microsoft', '1975-04-04','USA'),
 ('Apple', '1985-04-04','USA');

 select * from demo.delta.company_write_part;

INSERT overwrite table demo.delta.company_write_part
Partition(company_name = 'Microsoft') 
VALUES ('1975-04-04','USA') ;
 

 select * from demo.delta.company_write_part;

 desc history demo.delta.company_write_part
```
