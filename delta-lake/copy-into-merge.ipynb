Copy Into

```sql
create table if not exists demo.default.stock_price;

CREATE table if not exists demo.default.stock_raw

CREATE TABLE if not EXISTS demo.default.stock_target
```

```sql

COPY into demo.default.stock_raw 
from "abfss://demo@dbstoragecerttest.dfs.core.windows.net/stock"
FILEFORMAT= JSON
FORMAT_OPTIONS('interferSchema'= 'true')
COPY_OPTIONS ('mergeSchema'= 'true');
```

```sql
desc history demo.default.stock_price ;

RESTORE TABLE demo.default.stock_price to version as of 2;

select * from demo.default.stock_raw;

delete from demo.default.stock_raw where trading_date = '2025-02-11'
```

## Merge

```sql



MERGE INTO demo.default.stock_target as target using demo.default.stock_raw as source
on target.stock_id = source.stock_id
WHEN MATCHED and source.status = 'ACTIVE' then 
UPDATE SET target.price = source.price, target.trading_date = source.trading_date
WHEN MATCHED and source.status = 'DELISTED' then 
DELETE
When NOT MATCHED and source.status = 'ACTIVE' then 
INSERT(price, status, stock_id, trading_date) values(source.price, source.status, source.stock_id, source.trading_date)

```
