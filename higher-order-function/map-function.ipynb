## Map Functions 
- Transform_values
- Transform_keys
- Map_filter


```sql
CREATE or REPLACE view orders 
AS
SELECT * FROM VALUES
(1, map('smartphone',123, 'laptop',567, 'monitor', 456)),
(2, map('tablet', 435, 'headphone', 235, 'smartwatch',111)),
(3, map('keyboard', 500, 'mouse', 50))
AS orders(order_id, items) 

```

```sql
select * from orders
```

## Convert Map Keys to Upper

```sql
select order_id, TRANSFORM_KEYS(items, (item_name, item_price) -> UPPER(item_name)) as updated_items from orders;
```

## Apply 10% TAX on Price

```sql
select order_id, transform_values(items, (item_name, item_price) -> item_price + item_price*0.1) as tax_map from orders;
```

## Filter data which is having price more than 400

```sql
select order_id, MAP_FILTER(items, (key, price) -> price> 400)  as filtered_map from orders
```

```sql
select order_id, MAP_FILTER(items, (key, price) -> key like '%smart%') as filtered_map from orders
```
