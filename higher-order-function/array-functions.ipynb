Higher Order Functions
- Transform
- filter
- EXISTS
- Aggregate

```sql
CREATE OR Replace view order_items as 
select * from VALUES 
(1, array('smartphone', 'laptop', 'monitor')),
(2, array('tablet', 'headphone', 'smartwatch')),
(3, array('keyboard', 'mouse'))
as orders(order_id, items)
```

```sql
select * from order_items
```

## Transform - convert all array values to upper case

```sql
select order_id,transform(items, item -> UPPER(item)) as upper_array from order_items
```

## Filter data in array for smart

```sql
select order_id, FILTER(items, x -> x like '%smart%') as valid_array from order_items
```

## Check if specific value exists or not in array

```sql
select order_id, exists(items, item -> item = 'monitor') value_exists from order_items;

select order_id, exists(items, item -> item like '%smart%') value_exists from order_items
```

## Array with struct Type

```sql
CREATE OR Replace view order_items_complext as
select
  *
from
  VALUES
    (
      1,
      array(
        named_struct('name', 'smartphone', 'price', 123),
        named_struct('name', 'laptop', 'price', 453),
        named_struct('name', 'monitor', 'price', 234)
      )
    ),
    (
      2,
      array(
        named_struct('name', 'tablet', 'price', 123),
        named_struct('name', 'headphone', 'price', 453),
        named_struct('name', 'smartwatch', 'price', 234)
      )
    ),
    (
      3,
      array(
        named_struct('name', 'keyboard', 'price', 123), named_struct('name', 'mouse', 'price', 234)
      )
    )
    as orders (order_id, items)
```

```sql
select * from order_items_complext
```

## Transform struct array to convert values to upper and add 10% tax

```sql
select order_id, transform(items, item-> named_struct('name', upper(item.name),'price', (item.price + item.price*0.1))) as upper_array  from order_items_complext
```

## Aggregate price

```sql
select order_id, aggregate(items, 0, (acc, x)-> acc+x.price) as total_amount from order_items_complext
```
